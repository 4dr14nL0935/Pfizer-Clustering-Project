# EDA del Conjunto de Datos Original




```python
import pandas as pd
import numpy as np

# Cargar datos
df = pd.read_csv('VELSIPITY_AFF_MULTIPLI_prepared_prepared_prepared.csv')

print('Dimensiones del DataFrame:', df.shape)

df.head()
```

```python
# Tipos de datos de cada columna y número de valores no nulos
df_info = df.dtypes.to_frame(name='dtype')
df_info['non_null_count'] = df.notnull().sum()
df_info
```

```python
# Calcular y ordenar las columnas por porcentaje de valores faltantes
missing_counts = df.isnull().sum()
missing_ratio = (missing_counts / len(df)).sort_values(ascending=False)
missing_ratio = missing_ratio[missing_ratio > 0]

print('Columnas con valores faltantes y su porcentaje:')
print(missing_ratio)
```

```python
# Estadísticas descriptivas de columnas numéricas
numeric_df = df.select_dtypes(include=[np.number])
numeric_summary = numeric_df.describe().T

# Añadir asimetría y curtosis a la tabla
def skew_kurtosis(series):
    return pd.Series({'skew': series.skew(), 'kurtosis': series.kurtosis()})

skew_kurt = numeric_df.apply(skew_kurtosis).T

summary = numeric_summary.join(skew_kurt)
summary.sort_values(by='std', ascending=False).head(20)  # mostrar las 20 variables con mayor desviación estándar
```

```python
# Variables seleccionadas para analizar correlaciones principales
selected_cols = [col for col in ['UC_TRX','ORAL_TRX','IL23_TRX','BRAND1_TRX','BRAND2_TRX',
                 'UC_NRX','ORAL_NRX','IL23_NRX','BRAND1_NRX','BRAND2_NRX',
                 'UC_TRX_R4_16SUM','ORAL_NBRX_R4_29SUM','IL23_NBRX_R4_29SUM'] if col in df.columns]

corr_matrix = df[selected_cols].corr()

# Obtener pares de variables con mayor correlación absoluta (excepto la diagonal)
abs_corr = corr_matrix.abs().unstack()
abs_corr = abs_corr[abs_corr < 1]  # eliminar correlación de una variable consigo misma
sorted_pairs = abs_corr.sort_values(ascending=False).drop_duplicates()

# Mostrar las 10 correlaciones más altas
print('Las 10 correlaciones más altas entre variables seleccionadas:')
print(sorted_pairs.head(10))
```

```python
# Distribución de la variable de segmentación ATSEG
atseg_counts = df['ATSEG'].fillna('Missing').value_counts()

print('Distribución de ATSEG:')
print(atseg_counts)
```
